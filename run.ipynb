```python
filename = u.filenames(f'{path}')[0]
adjacency_list, num_seeds, round_type, graph_id = u.parse_JSON(filename)
G = nx.Graph(adjacency_list)

# Highest degree nodes
sorted_degree = sorted(dict(G.degree).items(), key=lambda x: x[1], reverse=True)
seeds = np.array(sorted_degree)[:int(num_seeds), 0]
print(f"Top {int(num_seeds)} nodes: " + str(seeds))
```

```python
# Round-Robin Strategy
if round_type == 'RR':
    start_time = time_sec = time.time()
    # num_seeds = 10, index = 5 ; num_seeds = 15, index = 3
    index = 5

    # 4 minutes test, 1 minute for submission
    while time.time() - start_time < 240 and index <= 8:
        print("Time passed: " + str(time.time()- start_time))
        highest_deg_nodes = np.array(sorted_degree)[:int(num_seeds) * index, 0]
        print(f"Processing best seeds from the top {int(num_seeds) * index} nodes: " + str(highest_deg_nodes))

        node_comp = {node: 0 for node in highest_deg_nodes}
        for i in range(len(highest_deg_nodes)-1):
            for j in range(i+1, len(highest_deg_nodes)):

                # time checks
                if time.time() - start_time > 240:
                    print("Time passed: " + str(time.time()- start_time))
                    break

                # uses sim.py to compare highest deg nodes
                result = u.compare_nodes(filename, highest_deg_nodes[i], highest_deg_nodes[j])
                node_comp[highest_deg_nodes[i]] += result
                node_comp[highest_deg_nodes[j]] += not result

        if time.time() - start_time <= 240:
            sorted_nodes = sorted(node_comp.items(), key=lambda x: x[1], reverse=True)
            seeds = np.array(sorted_nodes)[:int(num_seeds), 0]
            print(f"{index}: " + str(seeds))
            index += 1

# Jungle Strategy
else:
    highest_deg_nodes = np.array(sorted_degree)[int(num_seeds)*2:int(num_seeds)*6, 0]
    seeds = random.sample(sorted(highest_deg_nodes), int(num_seeds))
    print(seeds)
```

```python
u.generate_txt_files(path + 'output\\', round_type, num_seeds, graph_id, seeds)
u.download_files(path + 'output\\', seeds, round_type, num_seeds, graph_id)
```
